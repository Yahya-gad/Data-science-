```python
df['publish_time'] = pd.to_datetime(df['publish_time'])
df['trending_date']=pd.to_datetime(df['trending_date'],format= '%y.%d.%m')
```

```python
df.isna().sum()
```

```python
df.fillna(df['description'],inplace=True)
```

```python
df.drop(['video_id','tags','thumbnail_link','comments_disabled','ratings_disabled','video_error_or_removed','description'],
        axis=1,
        inplace=True)
```

```python
df.columns
```

```python
df.rename(columns={'channel_title':'channel',
                   'category_id':'category',
                   'comment_count':'comments'
                   

},inplace=True)
df
```

```python
df.drop_duplicates(inplace=True)
```

```python
df.dropna(inplace=True)
df
```

```python
df['published_date'] = df['publish_time'].dt.date
df['published_hour'] = df['publish_time'].dt.time
df['trending_year'] = df['trending_date'].dt.year
df['trending_month'] = df['trending_date'].dt.month
df['trending_day'] = df['trending_date'].dt.day
```

```python
df.drop(['published_date','trending_date'],
        axis=1,
        inplace=True)
```

```python
df['likes_ratio'] = df['likes'] / (df['likes'] + df['dislikes'])
```

```python
df
```

```python
mean_views=df['views'].mean()
mean_likes=df['likes'].mean()
mean_comments=df['comments'].mean()

print("mean_views:",mean_views)
print("mean_likes:",mean_likes)
print("mean_comments:",mean_comments)
```

```python
median_comments=df['comments'].median()
median_comments
```

```python
var_likes=df['likes'].var()
var_views=df['views'].var()
var_comments=df['comments'].var()
print("var_likes:",var_likes)
print("var_views:",var_views)
print("var_comments:",var_comments)
```

```python
std_likes=df['likes'].std()
std_views=df['views'].std()
print("std_likes:",std_likes)
print("std_views:",std_views)
```

```python
cov_matrix=df[['views','likes']].cov()
cov_matrix
```

```python
correlation=df[['views','likes']].corr()
correlation
```

```python
max_ratio=df['likes_ratio'].max()
max_ratio
```

```python
top_views_channel=df.groupby('channel')['views'].sum().idxmax()
top_likes_channel=df.groupby('channel')['likes'].sum().idxmax()
top_dislikes_channel=df.groupby('channel')['dislikes'].sum().idxmax()
print("top_views_category:",top_views_channel)
print("top_views_category:",top_views_channel)
print("top_dislikes_category:",top_dislikes_channel)
```

```python
category_counts=df['category'].value_counts()
plt.figure(figsize=(12,6))
category_counts.plot(kind='bar',color='skyblue')
plt.title('Number of Videos per Category')
plt.xlabel('category')
plt.ylabel('Number of Videos')
plt.xticks(rotation=45)
plt.grid(axis='y',linestyle='--',alpha=0.7)
plt.tight_layout()
plt.show()
```

Insight:
El videos el aktar mawgooda fe categories mo3ayana, da beyban en el nas bt7eb temshy 3ala mawade3 mo3ayana.


```python
#bar chart
top_views = df.sort_values('views', ascending=False).head(10)
plt.figure(figsize=(12,6))
plt.bar(top_views['title'], top_views['views'], color='purple')
plt.title('Top 10 Videos by Views')
plt.xlabel('Video Title')
plt.ylabel('Views')
plt.xticks(rotation=90)
plt.tight_layout()
plt.show()
```

Insight:
Fee videos mo3ayana mosta5dmeen shafouha aktar men ay 7aga tany, da y3ny enhom viral aw popular awy.


```python
#piechart
top_categories = df['channel'].value_counts().head(5)

plt.figure(figsize=(8, 8))
plt.pie(top_categories.values, labels=top_categories.index.astype(str),
        autopct='%1.1f%%', startangle=140)
plt.title('Top 5 Video Categories Distribution')
plt.legend(title='Category')
plt.axis('equal')
plt.tight_layout()
plt.show()
```

El 5 channels dol 3ndhom aktar 3adad videos fel dataset, we da yban enhom active awy we 3andhom content kteer.

```python
#histrogram
plt.figure(figsize=(10, 6))
plt.hist(df['views'], bins=30, color='skyblue', edgecolor='black')
plt.title('Distribution of Video Views')
plt.xlabel('Number of Views')
plt.ylabel('Frequency')
plt.legend(['Views'])
plt.tight_layout()
plt.show()
```

Insight:
Aktar el videos btgeeb views 2a2al, we fe videos 2aleela awy bas el views beta3ha 3aly geddan.

```python
#line graph
df['publish_time'] = pd.to_datetime(df['publish_time'])
df_sorted = df.sort_values(by='publish_time')
daily_views = df_sorted.groupby(df_sorted['publish_time'].dt.date)['views'].sum()

plt.figure(figsize=(12, 6))
plt.plot(daily_views.index, daily_views.values, color='green', linewidth=2)
plt.title('Total Views Over Time')
plt.xlabel('Publish Date')
plt.ylabel('Total Views')
plt.grid(True)
plt.legend(['Total Views'])
plt.tight_layout()
plt.show()
```

Insight:
Views btetghayar 3ala el wa2t, momken tekoun bteshtaghal 3ala mawsem aw trend mo3ayan.


```python
plt.figure(figsize=(12, 6))
data_to_plot = [df[df['category'] == cat]['views'] for cat in df['category'].unique()]
plt.boxplot(data_to_plot, labels=df['category'].unique())
plt.title('Box and Whiskers Plot of Views per Category')
plt.xlabel('Category')
plt.ylabel('Number of Views')
plt.grid(True)
plt.tight_layout()
plt.show()
```

insight:
Feh far2 kbeer ben el videos fi nafs el category, y3ny fee videos mosh mashya zay ba2y el videos
